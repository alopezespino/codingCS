# Python Fundamentals

**What you'll learn:** The building blocks of Python — variables, data types, lists, dictionaries, loops, functions, and a first look at NumPy and pandas.

**Prerequisites:** None. This is the starting point.

**Why Python?** Python is one of the most popular programming languages for data analysis. It has a clean, readable syntax that reads almost like English, a massive ecosystem of data-science libraries, and is used by companies like Google, Netflix, and NASA.

## 1. Variables and Assignment

A **variable** is a name that stores a value. Think of it as a labeled box — you put something in, and you can refer to it later by its label.

You create a variable with the `=` sign (called the **assignment operator**).

````python
x = 5
name = "Alice"
temperature = 72.3

print(x)
print(name)
print(temperature)
````

You can change the value of a variable at any time by assigning a new value to it:

````python
x = 5
print(x)

x = 10
print(x)
````

## 2. Data Types

Every value in Python has a **type** that determines what you can do with it. The four basic types are:

| Type | What it is | Example |
|------|------------|----------|
| `int` | Whole number | `42` |
| `float` | Decimal number | `3.14` |
| `str` | Text (string) | `"hello"` |
| `bool` | True or False | `True` |

Use the `type()` function to check what type a value is:

````python
print(type(42))
print(type(3.14))
print(type("hello"))
print(type(True))
````

````python
age = 25
price = 9.99
city = "Chicago"
is_student = True

print(type(age))
print(type(price))
print(type(city))
print(type(is_student))
````

## 3. Basic Arithmetic

Python works like a calculator. Here are the main arithmetic operators:

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | Addition | `3 + 2` | `5` |
| `-` | Subtraction | `3 - 2` | `1` |
| `*` | Multiplication | `3 * 2` | `6` |
| `/` | Division | `7 / 2` | `3.5` |
| `//` | Integer division | `7 // 2` | `3` |
| `%` | Remainder (modulo) | `7 % 2` | `1` |
| `**` | Exponent (power) | `3 ** 2` | `9` |

````python
print(3 + 2)
print(10 - 4)
print(6 * 7)
print(15 / 4)
print(15 // 4)
print(15 % 4)
print(2 ** 10)
````

````python
hours_worked = 40
hourly_rate = 25.50
weekly_pay = hours_worked * hourly_rate
print(f"Weekly pay: ${weekly_pay}")
````

## 4. Strings

A **string** is a sequence of characters (text). You create strings with quotes — single `'...'` or double `"..."` both work.

````python
greeting = "Hello"
target = "World"

# Concatenation: joining strings with +
message = greeting + ", " + target + "!"
print(message)
````

**f-strings** (formatted strings) are the modern way to insert variables into text. Put an `f` before the opening quote and use `{variable_name}` inside:

````python
name = "Alice"
age = 28
print(f"My name is {name} and I am {age} years old.")
print(f"Next year I will be {age + 1}.")
````

Strings have built-in **methods** — functions you call on the string with a dot:

````python
text = "  Hello, World!  "

print(text.upper())        # ALL CAPS
print(text.lower())        # all lowercase
print(text.strip())        # remove leading/trailing whitespace
print(text.split(","))     # split into a list at each comma
````

## 5. Lists

A **list** is an ordered collection of items. Lists are one of the most useful data structures in Python.

You create a list with square brackets `[]` and separate items with commas:

````python
fruits = ["apple", "banana", "cherry"]
numbers = [10, 20, 30, 40, 50]
mixed = [1, "hello", 3.14, True]

print(fruits)
print(numbers)
print(mixed)
````

**Indexing:** Access individual items by position. Python counts from 0, not 1.

```
Index:   0        1        2
List:  ["apple", "banana", "cherry"]
```

````python
print(fruits[0])    # first item
print(fruits[1])    # second item
print(fruits[-1])   # last item (negative index counts from the end)
````

**Slicing:** Get a sub-list with `[start:end]`. The start is included, the end is not.

````python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])   # items at index 1, 2, 3
print(numbers[:3])    # first 3 items
print(numbers[2:])    # from index 2 to the end
````

**Modifying lists:**

````python
fruits = ["apple", "banana", "cherry"]

fruits.append("date")       # add to the end
print(fruits)

print(len(fruits))          # how many items
````

## 6. Dictionaries

A **dictionary** stores data as **key-value pairs**. Think of it like a real dictionary: you look up a word (key) to find its definition (value).

You create a dictionary with curly braces `{}`:

````python
student = {
    "name": "Alice",
    "age": 22,
    "major": "Economics",
    "gpa": 3.8
}

print(student)
````

````python
# Access a value by its key
print(student["name"])
print(student["gpa"])

# Add a new key-value pair
student["university"] = "Purdue"
print(student)
````

## 7. Comparisons and Logic

**Comparison operators** compare two values and return `True` or `False`:

| Operator | Meaning |
|----------|----------|
| `==` | Equal to |
| `!=` | Not equal to |
| `<` | Less than |
| `>` | Greater than |
| `<=` | Less than or equal |
| `>=` | Greater than or equal |

````python
print(5 == 5)    # True
print(5 != 3)    # True
print(10 < 7)    # False
print(3 >= 3)    # True
````

**Logical operators** combine conditions:

- `and` — both must be True
- `or` — at least one must be True
- `not` — flips True to False and vice versa

````python
age = 25
has_id = True

print(age >= 21 and has_id)    # True — both conditions met
print(age < 18 or has_id)      # True — at least one is True
print(not has_id)              # False — flips True to False
````

## 8. Conditional Statements (if / elif / else)

Conditionals let your code make decisions. The structure is:

```python
if condition:
    # runs if condition is True
elif another_condition:
    # runs if the first was False and this one is True
else:
    # runs if none of the above were True
```

**Important:** The indented block (4 spaces) is how Python knows what code belongs to each branch.

````python
score = 85

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("F")
````

````python
temperature = 35

if temperature > 100:
    print("Boiling!")
elif temperature > 75:
    print("It's hot.")
elif temperature > 50:
    print("Nice weather.")
else:
    print("It's cold!")
````

## 9. Loops

Loops let you repeat code. The most common is the **for loop**, which goes through each item in a sequence.

````python
# Loop over a list
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(f"I like {fruit}")
````

The `range()` function generates a sequence of numbers. `range(5)` gives you 0, 1, 2, 3, 4:

````python
for i in range(5):
    print(f"Step {i}")
````

````python
# A practical example: summing a list of numbers
prices = [12.50, 8.99, 25.00, 3.75]
total = 0

for price in prices:
    total = total + price

print(f"Total: ${total:.2f}")
````

## 10. Functions

A **function** is a reusable block of code that performs a specific task. You define a function with the `def` keyword:

```python
def function_name(parameters):
    # code that runs when the function is called
    return result
```

````python
def greet(name):
    return f"Hello, {name}!"

message = greet("Alice")
print(message)
print(greet("Bob"))
````

````python
def calculate_tax(price, tax_rate=0.07):
    """Calculate the total price including tax."""
    tax = price * tax_rate
    return price + tax

print(calculate_tax(100))          # uses default 7% rate
print(calculate_tax(100, 0.10))    # uses 10% rate
````

## 11. Importing Libraries

A **library** (also called a **package**) is a collection of pre-written code that adds new capabilities. Instead of writing everything from scratch, you import libraries that others have built.

The `import` statement loads a library so you can use it:

```python
import numpy as np      # import numpy, refer to it as "np" for short
import pandas as pd     # import pandas, refer to it as "pd" for short
```

The `as np` / `as pd` parts are standard short names that everyone uses.

## 12. NumPy Basics

**NumPy** (Numerical Python) is the foundation for numerical computing in Python. Its core feature is the **array** — a fast, efficient way to work with lists of numbers.

````python
import numpy as np

# Create an array from a list
numbers = np.array([10, 20, 30, 40, 50])
print(numbers)
print(type(numbers))
````

The big advantage of NumPy arrays: you can do math on all elements at once (this is called **vectorized operations**):

````python
numbers = np.array([10, 20, 30, 40, 50])

print(numbers * 2)       # multiply every element by 2
print(numbers + 100)     # add 100 to every element
print(numbers ** 2)      # square every element
````

````python
# Useful array functions
print(f"Mean: {numbers.mean()}")
print(f"Sum: {numbers.sum()}")
print(f"Min: {numbers.min()}")
print(f"Max: {numbers.max()}")
print(f"Shape: {numbers.shape}")  # shape tells you the dimensions
````

## 13. Pandas: Your Data Analysis Toolkit

**pandas** is the go-to library for working with tabular data (rows and columns, like a spreadsheet). Its core object is the **DataFrame** — a table.

Let's load a real dataset from a CSV file:

````python
import pandas as pd

# Read a CSV file into a DataFrame
df = pd.read_csv("../data/employees.csv")

# View the first 5 rows
df.head()
````

````python
# How many rows and columns?
print(f"Shape: {df.shape}")
print(f"That's {df.shape[0]} rows and {df.shape[1]} columns.")
````

````python
# What are the column names and their types?
df.dtypes
````

````python
# Summary statistics for numeric columns
df.describe()
````

### Selecting columns

Use square brackets with the column name to select a single column:

````python
# Single column — returns a Series
salaries = df["salary"]
print(salaries.head())
print(f"\nAverage salary: ${salaries.mean():,.2f}")
````

````python
# Multiple columns — returns a DataFrame
subset = df[["name", "department", "salary"]]
subset.head()
````

### Filtering rows

You can filter a DataFrame using a condition inside square brackets:

````python
# Employees with salary above 100,000
high_earners = df[df["salary"] > 100000]
print(f"Found {len(high_earners)} employees earning over $100k:")
high_earners[["name", "department", "salary"]]
````

````python
# Combine conditions with & (and) or | (or)
engineering_senior = df[(df["department"] == "Engineering") & (df["age"] > 40)]
engineering_senior[["name", "department", "age", "salary"]]
````

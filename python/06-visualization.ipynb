# Data Visualization with Python

**What you'll learn:** How to create informative charts using matplotlib and seaborn — line plots, bar charts, scatter plots, histograms, box plots, heatmaps, and more.

**Prerequisites:** [01 - Python Basics](01-python-basics.ipynb)

## Why Visualize Data?

Numbers alone can be misleading. A table of averages might hide that one group has huge outliers. A trend might only be obvious when plotted on a timeline. Visualization helps you:

- **Explore** — spot patterns, outliers, and relationships quickly
- **Communicate** — a good chart tells a story faster than a paragraph of numbers
- **Verify** — check whether your analysis makes sense before drawing conclusions

## Setup

We'll use two libraries:

- **matplotlib** — the foundational plotting library in Python. It gives you full control.
- **seaborn** — built on top of matplotlib. It makes statistical plots easier and prettier.

The `%matplotlib inline` line tells Jupyter to display plots directly in the notebook.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

%matplotlib inline

# Load our datasets
employees = pd.read_csv("../data/employees.csv")
sales = pd.read_csv("../data/sales.csv")
```

```python
# Quick look at the data
print("Employees:")
print(employees.head(3))
print(f"\nSales:")
print(sales.head(3))
```

## matplotlib Basics

### Line Plot

Line plots show trends over a sequence (often time). Let's plot monthly sales revenue:

```python
# Prepare data: total revenue per month
sales["date"] = pd.to_datetime(sales["date"])
sales["month"] = sales["date"].dt.to_period("M")
monthly = sales.groupby("month").apply(
    lambda x: (x["quantity"] * x["unit_price"]).sum()
).reset_index()
monthly.columns = ["month", "revenue"]
monthly["month_str"] = monthly["month"].astype(str)

# Plot
plt.figure(figsize=(10, 4))
plt.plot(monthly["month_str"], monthly["revenue"], marker="o")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.title("Monthly Sales Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Key functions:**
- `plt.figure(figsize=(width, height))` — set the plot size in inches
- `plt.plot(x, y)` — draw a line
- `plt.xlabel()`, `plt.ylabel()`, `plt.title()` — add labels
- `plt.show()` — display the plot

### Bar Chart

Bar charts compare quantities across categories:

```python
# Number of employees per department
dept_counts = employees["department"].value_counts()

plt.figure(figsize=(8, 4))
plt.bar(dept_counts.index, dept_counts.values, color="steelblue")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.title("Employees by Department")
plt.tight_layout()
plt.show()
```

### Scatter Plot

Scatter plots show the relationship between two numeric variables:

```python
plt.figure(figsize=(8, 5))
plt.scatter(employees["age"], employees["salary"], alpha=0.6, color="teal")
plt.xlabel("Age")
plt.ylabel("Salary ($)")
plt.title("Age vs Salary")
plt.tight_layout()
plt.show()
```

`alpha=0.6` makes dots slightly transparent, which helps when points overlap.

### Histogram

Histograms show the **distribution** of a single numeric variable — how values are spread out:

```python
plt.figure(figsize=(8, 4))
plt.hist(employees["salary"], bins=15, color="coral", edgecolor="white")
plt.xlabel("Salary ($)")
plt.ylabel("Count")
plt.title("Salary Distribution")
plt.tight_layout()
plt.show()
```

`bins=15` controls how many bars the histogram has. More bins = more detail.

### Adding a Legend

When you have multiple series on one plot, add a **legend** to tell them apart:

```python
sales["revenue"] = sales["quantity"] * sales["unit_price"]
elec = sales[sales["category"] == "Electronics"].groupby("month")["revenue"].sum()
cloth = sales[sales["category"] == "Clothing"].groupby("month")["revenue"].sum()

plt.figure(figsize=(10, 4))
plt.plot(elec.index.astype(str), elec.values, marker="o", label="Electronics")
plt.plot(cloth.index.astype(str), cloth.values, marker="s", label="Clothing")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.title("Revenue by Category")
plt.legend()  # shows the legend
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## seaborn: Statistical Visualization

seaborn builds on matplotlib and provides higher-level functions for common statistical plots. The defaults look nicer too.

### Box Plot

A **box plot** shows the distribution of a numeric variable, split by a category. The box shows the middle 50% of values; the line inside is the median; the whiskers extend to the rest; dots are outliers.

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=employees, x="department", y="salary", palette="Set2")
plt.title("Salary Distribution by Department")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### Heatmap

A **heatmap** uses color to show the magnitude of values in a grid. It's great for visualizing correlations:

```python
# Correlation matrix of numeric columns
numeric_cols = employees[["salary", "age"]]
corr = numeric_cols.corr()

plt.figure(figsize=(5, 4))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

`annot=True` prints the numbers on each cell. `cmap` sets the color scheme.

### Count Plot

Like a bar chart, but seaborn counts the categories for you:

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=sales, x="category", palette="pastel",
              order=sales["category"].value_counts().index)
plt.title("Number of Transactions by Category")
plt.tight_layout()
plt.show()
```

### Pair Plot

A **pair plot** shows scatter plots for every pair of numeric columns. It's a quick way to explore relationships in a dataset:

```python
sns.pairplot(employees[["salary", "age", "department"]], hue="department", height=2.5)
plt.suptitle("Employee Attributes by Department", y=1.02)
plt.show()
```

`hue="department"` colors dots by department so you can see group differences.

### Customizing Color Palettes and Figure Size

```python
# seaborn has many built-in palettes
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.countplot(data=sales, x="category", palette="Set1", ax=axes[0])
axes[0].set_title("Set1")

sns.countplot(data=sales, x="category", palette="viridis", ax=axes[1])
axes[1].set_title("viridis")

sns.countplot(data=sales, x="category", palette="muted", ax=axes[2])
axes[2].set_title("muted")

for ax in axes:
    ax.tick_params(axis="x", rotation=30)

plt.tight_layout()
plt.show()
```

## What Is a Data Analysis Report?

A data analysis report tells a story with data. It typically has:

1. **Introduction** — what question are we answering?
2. **Data description** — where did the data come from? What does it contain?
3. **Analysis** — the actual exploration, calculations, and charts
4. **Findings** — what did we discover?
5. **Conclusion** — what do we recommend?

Jupyter notebooks are ideal for this because they combine text, code, and output in a single document that anyone can read (and reproduce).

## Summary Statistics with pandas

Reports almost always include summary statistics. Here are the most useful pandas tools for creating them.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

sales = pd.read_csv("../data/sales.csv")
sales["date"] = pd.to_datetime(sales["date"])
sales["revenue"] = sales["quantity"] * sales["unit_price"]

employees = pd.read_csv("../data/employees.csv")
````

### .describe() — Quick overview of numeric columns

````python
sales[["quantity", "unit_price", "revenue"]].describe().round(2)
````

### .value_counts() — Frequency of categories

````python
sales["category"].value_counts()
````

### .groupby().agg() — Custom summary tables

````python
summary = sales.groupby("category").agg(
    transactions=("transaction_id", "count"),
    total_revenue=("revenue", "sum"),
    avg_price=("unit_price", "mean"),
    avg_quantity=("quantity", "mean")
).round(2)

summary = summary.sort_values("total_revenue", ascending=False)
summary
````

### pd.crosstab() — Cross-tabulation of two categories

````python
# How many transactions per category per region?
ct = pd.crosstab(sales["category"], sales["region"])
ct
````

## Formatting pandas Output

### .style for highlighting

````python
# Highlight the max value in each column
summary.style.highlight_max(axis=0, color="lightgreen")
````

````python
# Format numbers with dollar signs and commas
summary.style.format({
    "total_revenue": "${:,.0f}",
    "avg_price": "${:,.2f}",
    "avg_quantity": "{:.1f}"
})
````

### .to_markdown() for clean text tables

If you need the table as plain text (for pasting into emails or documents):

````python
print(summary.to_markdown())
````

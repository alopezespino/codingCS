## What is R?

R is a programming language designed for statistics and data analysis. It was created by statisticians, so it has built-in support for things like calculating averages, running regressions, and making charts. R is one of the two most popular languages for data work (the other being Python — you'll learn that too).

R is especially popular in academia, healthcare, finance, and any field that relies on statistics.

## Variables and Assignment

A **variable** is a name that stores a value. Think of it like a labeled box — you put something inside and can refer to it later by name.

In R, the standard way to assign a value to a variable is with the arrow `<-`. You can also use `=`, but `<-` is the R convention.

````r
# Assign the value 10 to a variable called x
x <- 10

# Print the value of x (in Jupyter, just typing the variable name prints it)
x
````

````r
# You can use = too, but <- is preferred in R
y = 20
y
````

````r
# Variables can store text (called "character strings" in R)
greeting <- "Hello, world!"
greeting
````

## Data Types

Every value in R has a **type**. The three most common types are:

| Type | What it stores | Examples |

| `character` | Text (always in quotes) | `"hello"`, `"R is fun"` |

Use `class()` to check what type a value is.

````r
class(42)
````

````r
class("hello")
````

````r
class(TRUE)
````

## Basic Arithmetic

R works like a calculator. The standard math operators are:

| Operator | Meaning | Example |
|----------|---------|--------|
| `+` | Addition | `5 + 3` → 8 |
| `-` | Subtraction | `10 - 4` → 6 |

## Vectors

A **vector** is R's most basic data structure — it's an ordered list of values of the same type. You create a vector with the `c()` function (short for "combine" or "concatenate").

**Important:** R uses 1-based indexing. The first element is at position 1 (not 0 like in Python).

````r
# Create a vector of temperatures
temps <- c(72, 68, 75, 80, 77, 69, 73)
temps
````

````r
# Access individual elements (1-based!)
temps[1]   # first element
````

````r
temps[3]   # third element
````

````r
# Get a range of elements
temps[2:4]   # elements 2 through 4
````

### Vectorized Operations

One of R's strengths is that math operations work on entire vectors at once — no need to loop through each element.

````r
# Convert all temperatures from Fahrenheit to Celsius
(temps - 32) * 5 / 9
````

````r
# Useful built-in functions for vectors
length(temps)   # how many elements
````

````r
sum(temps)      # add them all up
````

````r
mean(temps)     # average
````

````r
min(temps)      # smallest value
````

````r
max(temps)      # largest value
````

## Data Frames

A **data frame** is R's version of a spreadsheet — a table with rows and columns. Each column is a vector, and columns can have different types (one column might be numbers, another text).

You create one with `data.frame()` and access columns with `$`.

````r
# Create a small data frame
students <- data.frame(
  name = c("Alice", "Bob", "Carlos"),
  grade = c(92, 85, 78),
  passed = c(TRUE, TRUE, TRUE)
)

students
````

````r
# Access a single column with $
students$grade
````

````r
# Calculate the average grade
mean(students$grade)
````

## The Pipe Operator: `|>`

The **pipe** takes the result of whatever is on its left and passes it as the first argument to the function on its right. Read it as **"and then"**.

R has two pipe operators:
- `|>` — the built-in pipe (R 4.1+)
- `%>%` — the tidyverse pipe (from the `magrittr` package)

They do the same thing. We'll use `|>` since it's built in.

Without the pipe, you nest function calls inside each other, which is hard to read:
```r
round(mean(c(1, 2, 3, 4, 5)), 2)
```

With the pipe, you read left to right:
```r
c(1, 2, 3, 4, 5) |> mean() |> round(2)
```

````r
# Without pipe (read inside-out)
round(mean(c(12.3, 45.6, 78.9)), 1)
````

````r
# With pipe (read left to right: create vector, THEN take mean, THEN round)
c(12.3, 45.6, 78.9) |> mean() |> round(1)
````

## dplyr: Transforming Data

The `dplyr` package gives you five main "verbs" for working with data. Each one does one thing:

| Verb | What it does | SQL equivalent |
|------|-------------|----------------|
| `select()` | Pick columns | SELECT |
| `filter()` | Pick rows based on conditions | WHERE |
| `mutate()` | Create or modify columns | (computed columns) |
| `arrange()` | Sort rows | ORDER BY |
| `summarize()` | Calculate summary statistics | aggregate functions |

You chain them together with the pipe `|>`. Let's try each one.

### select() — Choose Columns

`select()` keeps only the columns you name.

````r
# Keep only name, department, and salary
employees |> select(name, department, salary)
````

````r
# Drop a column with a minus sign
employees |> select(-id)
````

### filter() — Choose Rows

`filter()` keeps only the rows where a condition is TRUE.

````r
# Employees in the Engineering department
employees |> filter(department == "Engineering")
````

````r
# Employees earning more than 100,000
employees |> filter(salary > 100000)
````

````r
# Combine conditions: Engineering AND salary above 80,000
employees |> filter(department == "Engineering", salary > 80000)
````

### mutate() — Create New Columns

`mutate()` adds new columns (or modifies existing ones) based on calculations.

````r
# Add a column for monthly salary
employees |> mutate(monthly_salary = salary / 12)
````

````r
# Add a column that labels salary level
employees |> mutate(
  salary_level = if_else(salary > 100000, "high", "standard")
)
````

### arrange() — Sort Rows

`arrange()` sorts the data. Use `desc()` for descending order.

````r
# Sort by salary (lowest first)
employees |> arrange(salary)
````

````r
# Sort by salary (highest first)
employees |> arrange(desc(salary))
````

### summarize() + group_by() — Aggregate Data

`summarize()` collapses the data into summary statistics. Combined with `group_by()`, it computes those statistics per group.

````r
# Overall average salary
employees |> summarize(avg_salary = mean(salary))
````

````r
# Average salary BY department
employees |>
  group_by(department) |>
  summarize(
    avg_salary = mean(salary),
    count = n()
  ) |>
  arrange(desc(avg_salary))
````

### Chaining It All Together

The real power comes from chaining multiple verbs with `|>`. Read this as a sentence: "Take employees, **then** filter to Sales and Marketing, **then** select name and salary, **then** arrange by salary descending."

````r
employees |>
  filter(department %in% c("Sales", "Marketing")) |>
  select(name, department, salary) |>
  arrange(desc(salary))
````
